### **1. Government Expenditures**

```python
# STEP 1: Upload file
from google.colab import files
uploaded = files.upload()  # Upload 'Government expenditures.csv'

# STEP 2: Install dependencies
!pip install statsmodels

# STEP 3: Run data cleaning
import pandas as pd
import numpy as np
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# Replace with your actual filename
df = pd.read_csv('Government expenditures.csv')

# Clean and filter
df = df.drop(columns=['Unnamed: 4', 'Unnamed: 5'], errors='ignore')
df = df[df['GovernmentExpenditure_Label'] == 'Total function']

# 34 OECD countries
oecd_34 = [
    'Canada', 'Mexico', 'United States', 'Austria', 'Belgium', 'France', 'Germany',
    'Ireland', 'Luxembourg', 'Netherlands', 'Switzerland', 'United Kingdom', 'Denmark',
    'Estonia', 'Finland', 'Iceland', 'Latvia', 'Lithuania', 'Norway', 'Sweden', 'Greece',
    'Italy', 'Portugal', 'Spain', 'Turkey', 'Czech Republic', 'Hungary', 'Poland',
    'Slovak Republic', 'Slovenia', 'Australia', 'Japan', 'Korea', 'New Zealand'
]

df = df[df['Economy_Label'].isin(oecd_34)]

# Pivot to time series
pivot_df = df.pivot(index='Year', columns='Economy_Label', values='US_dollars_at_current_prices_in_millions_Value')
pivot_df = pivot_df.sort_index()

# Extend years to 2023
all_years = list(range(pivot_df.index.min(), 2024))
pivot_df = pivot_df.reindex(all_years)

# Forecast and backfill
forecasted_df = pivot_df.copy()

for country in forecasted_df.columns:
    series = forecasted_df[country]

    # Forecast 2020–2023
    forward_series = series.dropna()
    if forward_series.index.max() < 2023 and len(forward_series) > 5:
        try:
            model = ExponentialSmoothing(forward_series, trend="add", seasonal=None, initialization_method="estimated")
            fit = model.fit()
            forecast = fit.predict(start=2020, end=2023)
            forecasted_df.loc[2020:2023, country] = forecast
        except:
            pass

    # Backfill 2000–2010
    if series.isnull().sum() > 0:
        reversed_series = series[::-1].dropna()
        if len(reversed_series) > 5:
            try:
                model = ExponentialSmoothing(reversed_series, trend="add", seasonal=None, initialization_method="estimated")
                fit = model.fit()
                forecast = fit.predict(start=0, end=len(series) - len(reversed_series) - 1)
                forecast = forecast[::-1]
                forecast.index = sorted(series.index[series.isnull()])
                for i, year in enumerate(forecast.index):
                    forecasted_df.at[year, country] = forecast.iloc[i]
            except:
                pass

# Fill remaining missing values
forecasted_df = forecasted_df.ffill()

# Export to Excel
forecasted_df.to_excel('Fully_Cleaned_Government_Expenditures_Final.xlsx')

# Download file to local machine
files.download('Fully_Cleaned_Government_Expenditures_Final.xlsx')
```
